## 4. Start the workspace

Open the ngrok URL printed by the next cell to access video translator.


```python
import subprocess
import threading
import sys
from pyngrok import ngrok
from rich import print as rprint
from rich.panel import Panel

def print_output(process):
    for line in iter(process.stdout.readline, ''):
        sys.stdout.write(line)
    for line in iter(process.stderr.readline, ''):
        sys.stderr.write(line)

# Start web interface
webui_process = subprocess.Popen(
    [sys.executable, "run_webui.py", "--host", "0.0.0.0"],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    universal_newlines=True,
    bufsize=1
)

# Create and start the output printing thread
output_thread = threading.Thread(target=print_output, args=(webui_process,))
output_thread.start()

# Create a tunnel using ngrok
public_url = ngrok.connect(8501)
rprint(Panel(f"web interface is available at Ngrok ⬇️", expand=False))
print(f"Click 👉 {public_url}")

# Keep the program running
ngrok_process = ngrok.get_ngrok_process()
try:
    webui_process.wait()
except KeyboardInterrupt:
    print("Interrupted by user, shutting down...")
finally:
    ngrok.kill()
    webui_process.terminate()
    output_thread.join()
```
